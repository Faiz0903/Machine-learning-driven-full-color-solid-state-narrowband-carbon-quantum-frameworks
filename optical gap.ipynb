```python
import pandas as pd
origin_data=pd.read_excel('trimer_gap.xlsx',sheet_name='new molecules predictied')
bridge_data=pd.read_excel('trimer_gap.xlsx',sheet_name='fsub')
final_data=origin_data.iloc[:,5:7]
final_data.iloc[:,1]=final_data.iloc[:,1]/1.32
for i in range(len(final_data)):
    if origin_data.iloc[i,2]<18 and origin_data.iloc[i,2]>4:
        final_data.iloc[i,1]=final_data.iloc[i,1]/1.09
for j in range(len(final_data)):
    if origin_data.iloc[j,2]>=18:
        k=origin_data.iloc[j,2]-17
        final_data.iloc[j,1]=final_data.iloc[j,1]/bridge_data.iloc[k-1,6]
        final_data.iloc[j,1]=final_data.iloc[j,1]/bridge_data.iloc[k-1,14]
        final_data.iloc[j,1]=final_data.iloc[j,1]/1.03
        final_data.iloc[j,1]=final_data.iloc[j,1]/1.07
final_data['wavelength']=1240/final_data.iloc[:,1]
data=pd.concat([origin_data.iloc[:,0],final_data],axis=1)
from decimal import Decimal, ROUND_HALF_UP
gap_column = final_data.columns[1]
data[gap_column] = data[gap_column].map(
    lambda value: float(Decimal(str(value)).quantize(Decimal('0.0001'), rounding=ROUND_HALF_UP))
)
data.to_excel('optical_gap.xlsx')
data
```
