```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import os

# Set the global font to Arial, size 18
plt.rcParams["font.family"] = ["Arial", "Microsoft YaHei", "SimHei"]
plt.rcParams['font.size'] = 18
plt.rcParams['axes.labelsize'] = 18
plt.rcParams['axes.titlesize'] = 18
plt.rcParams['xtick.labelsize'] = 16
plt.rcParams['ytick.labelsize'] = 16
plt.rcParams['legend.fontsize'] = 16
plt.rcParams['figure.titlesize'] = 18
plt.rcParams['axes.unicode_minus'] = False  # Display minus signs correctly

# Read the Excel file
desktop = Path.cwd()
data_path = os.path.join(desktop, "gs-homo-lumo-features.xlsx")

# Create the output folder
output_folder = desktop / 'correlation_analysis_results'
output_folder.mkdir(exist_ok=True)
print(f"Output folder: {output_folder}")

# Check whether the file exists
if not os.path.exists(data_path):
    print(f"File does not exist: {data_path}")
    print(f"Desktop path: {desktop}")
    print("\nExcel files on the desktop:")
    excel_files = list(desktop.glob('*.xlsx')) + list(desktop.glob('*.xls'))
    for f in excel_files:
        print(f"  - {f.name}")
    raise FileNotFoundError(f"Please ensure the file 'gs-homo-lumo-features.xlsx' is on the desktop")

print(f"Reading file: {data_path}")
df = pd.read_excel(data_path, index_col=0)  # Use the first column as the index (Index)

print(f"\nFile read successfully!")
print("Data shape:", df.shape)
print("Column names:", df.columns.tolist())
print("\nFirst 5 rows:")
print(df.head())

# Use the energy gap as the target variable
target_col = 'ΔEH-L'

# Calculate Pearson correlation coefficients
correlations = df.corr()[target_col].drop(target_col)

# Sort by absolute value and select the top 29 most correlated features
top_13_features = correlations.abs().nlargest(29).index
top_13_correlations = correlations[top_13_features]

print(f"\nWith '{target_col}'Top 29 most correlated features:")
for i, (feature, corr) in enumerate(top_13_correlations.items(), 1):
    print(f"{i:2d}. {feature}: {corr:.4f}")

# Extract data for the top 13 features
top_13_data = df[top_13_features.tolist() + [target_col]]

# Calculate the correlation matrix for these 13 features
corr_matrix = top_13_data.corr()

# 1. Plot the heatmap
fig, ax = plt.subplots(figsize=(16, 14))

# Create the heatmap
sns.heatmap(corr_matrix, 
            annot=True,  # Show values
            fmt='.2f',   # Number format
            cmap='coolwarm',  # Color map
            center=0,    # Center the color scale at 0
            square=True,  # Square cells
            linewidths=0.5,  # Cell border line width
            cbar_kws={"shrink": 0.8, "label": f"Pearson correlation coefficient"},
            ax=ax,
            annot_kws={'size': 10})  # Font size of numbers in the heatmap

target_label = r'$\Delta E_{\mathrm{H-L}}$'
ax.set_xticklabels([target_label if label.get_text() == target_col else label.get_text() for label in ax.get_xticklabels()])
ax.set_yticklabels([target_label if label.get_text() == target_col else label.get_text() for label in ax.get_yticklabels()])

# Set the title and labels
ax.set_title(f'Top 29 features vs. {target_label}: Pearson correlation heatmap', fontsize=18, pad=20, fontweight='bold')

# Rotate labels to avoid overlap
plt.xticks(rotation=45, ha='right', fontsize=16)
plt.yticks(rotation=0, fontsize=16)

# Adjust the layout
plt.tight_layout()

# Save the image
save_path = output_folder / '1_correlation_heatmap_top29.png'
plt.savefig(save_path, dpi=300, bbox_inches='tight')
print(f"\nHeatmap saved to: {save_path}")
plt.close()  # Close the figure to prevent display
```
